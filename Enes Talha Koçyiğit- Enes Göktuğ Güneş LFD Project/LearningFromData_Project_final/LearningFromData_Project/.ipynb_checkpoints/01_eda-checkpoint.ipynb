```python
# News Text Classification Project

## Objective The goal of this project is to classify news articles based on their textual content using machine learning and deep learning approaches.
```

```python
import pandas as pd
import re
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Reading and cleaning part of csv file 
df = pd.read_csv("data.csv")
df = df[['text', 'label']]

print(f"Before cleaning: {df.shape}")

# cleaning lack columns
df = df.dropna(subset=['text', 'label'])

# Removing duplicates
df = df.drop_duplicates()

# Ensure text is string
df['text'] = df['text'].astype(str)

# Text cleaning function
def clean_text(text):
    text = text.lower()
    text = re.sub(r"http\S+", "", text)
    text = re.sub(r"[^a-zA-ZğüşöçıİĞÜŞÖÇ ]", " ", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

# Last null check
df.isnull().sum()
df['text'] = df['text'].apply(clean_text)

# Remove empty texts after cleaning
df = df[df['text'].str.len() > 0]

# Remove very short texts, they might be dirty data
df['text_length'] = df['text'].apply(lambda x: len(x.split()))
df = df[df['text_length'] >= 20]

LEAKAGE_WORDS = [
    "sputnik", "sputniknews", "rossiya", "segodnya", "mia",
    "gzt", "jurnal", "arkitekt", "zpor", "gztmzt",
    "snob", "magazin", "magazi",
    "dh", "donanımhaber", "teknoloji",
    "anasayfa", "iletişim", "künye", "bize ulaşın",
    "reklam", "bildirim", "üye", "giriş",
    "spor", "futbol", "basketbol"
]

def clean_leakage(text):
    if not isinstance(text, str):
        return ""

    #  Önce senin standart temizliğin
    text = clean_text(text)

    #  Leakage kelimelerini temizle
    for word in LEAKAGE_WORDS:
        pattern = r'\b' + re.escape(word) + r'\b'
        text = re.sub(pattern, '', text)

    #  Fazla boşlukları tekrar temizle
    text = re.sub(r'\s+', ' ', text).strip()
    return text

```

```python
#After cleaning null check for each row and their total number
print(f"After cleaning: {df.shape}")

# Final null check
df.isnull().sum()

df.to_csv("clean_news_dataset.csv", index=False) #save clean data as new csv
```
